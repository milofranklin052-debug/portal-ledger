```python
import os, requests, pandas as pd
from dotenv import load_dotenv

load_dotenv()
headers = {"Authorization": f"Bearer {os.getenv('CFBD_KEY')}"}
```

```python
url = "https://api.collegefootballdata.com/stats/player/season"
params = {"year": 2024, "conference": "SEC"}
r = requests.get(url, headers=headers, params=params)
print(r.status_code)
print(r.text[:300])
```

```python
stats = pd.DataFrame(r.json())
print(stats.shape)
stats.head(10)
```

```python
wr = stats[(stats["category"] == "receiving") & (stats["statType"] == "YDS")]
wr.sort_values("stat", ascending=False).head(10)[["player", "team", "stat"]]
```

```python
stats["stat"] = pd.to_numeric(stats["stat"], errors="coerce")
print(stats["stat"].dtype)
```

```python
wr = stats[(stats["category"] == "receiving") & (stats["statType"] == "YDS")]
wr.sort_values("stat", ascending=False).head(10)[["player", "team", "stat"]]
```

```python
wide = stats.pivot_table(
    index=["playerId", "player", "team", "position"],
    columns=["category", "statType"],
    values="stat",
    aggfunc="first"
)
wide.columns = [f"{cat}_{stat}" for cat, stat in wide.columns]
wide = wide.reset_index()
print(wide.shape)
wide.head()
```

```python
wide[["player", "team", "position", "receiving_REC", "receiving_YDS", "receiving_TD"]] \
    .sort_values("receiving_YDS", ascending=False).head(10)
```

```python
r = requests.get("https://api.collegefootballdata.com/player/portal",
                 headers=headers, params={"year": 2024})
print(r.status_code)
portal = pd.DataFrame(r.json())

portal["name"] = (portal["firstName"].str.strip() + " " + portal["lastName"].str.strip()).str.lower()
wide["name"] = wide["player"].str.strip().str.lower()
```

```python
m = portal.merge(
    wide,
    left_on=["name", "destination"],
    right_on=["name", "team"],
    how="left",
    indicator=True
)
print(m["_merge"].value_counts())
```

```python
dest = m[m["destination"].isin(wide["team"].unique())]
print(dest["_merge"].value_counts())
miss = dest[dest["_merge"] == "left_only"][["name", "destination", "position_x", "stars", "eligibility"]]
miss.head(20)
```

```python
misses = dest[dest["_merge"] == "left_only"].copy()

def classify(row):
    pool = wide[wide["team"] == row["destination"]]
    last = row["lastName"].strip().lower()
    hits = pool[pool["name"].str.contains(last, regex=False)]
    return "possible name mismatch" if len(hits) > 0 else "no stats found"

misses["cause"] = misses.apply(classify, axis=1)
print(misses["cause"].value_counts())
```

```python
look = misses[misses["cause"] == "possible name mismatch"]
for _, row in look.head(10).iterrows():
    pool = wide[wide["team"] == row["destination"]]
    last = row["lastName"].strip().lower()
    print(row["name"], "->", pool[pool["name"].str.contains(last, regex=False)]["name"].tolist())
```

```python
import re

def clean(name):
    name = name.lower()
    name = re.sub(r"[.\']", "", name)                   # remove periods and apostrophes
    name = re.sub(r"\b(jr|sr|ii|iii|iv|v)\b", "", name) # remove suffixes
    return re.sub(r"\s+", " ", name).strip()            # collapse extra spaces

portal["name_clean"] = portal["name"].apply(clean)
wide["name_clean"] = wide["name"].apply(clean)
```

```python
m2 = portal.merge(wide, left_on=["name_clean", "destination"],
                  right_on=["name_clean", "team"], how="left", indicator=True)
dest2 = m2[m2["destination"].isin(wide["team"].unique())]
print(dest2["_merge"].value_counts())
```

```python
from difflib import SequenceMatcher

def sim(a, b):
    return SequenceMatcher(None, a, b).ratio()
```

```python
print(sim("vandrevius", "vandrevious"))
print(sim("charlie", "chaz"))
print(sim("benjamin", "ben"))
```

```python
def fuzzy_match(row):
    pool = wide[wide["team"] == row["destination"]]
    parts = row["name_clean"].split()
    first, last = parts[0], parts[-1]
    best, best_score = None, 0
    for _, p in pool.iterrows():
        pp = p["name_clean"].split()
        if pp[-1] != last:          # last name must match exactly
            continue
        score = sim(first, pp[0])
        if pp[0].startswith(first) or first.startswith(pp[0]):
            score = max(score, 0.85)   # catches ben/benjamin
        if score > best_score:
            best, best_score = p["name_clean"], score
    return pd.Series([best, best_score])

un = m2[(m2["_merge"] == "left_only") & m2["destination"].isin(wide["team"].unique())].copy()
un[["candidate", "score"]] = un.apply(fuzzy_match, axis=1)
print(un[un["score"] > 0][["name_clean", "candidate", "score"]].sort_values("score", ascending=False))
```

```python
accepted = {
    "vandrevius jacobs": "vandrevious jacobs",
    "tiaoalii savea": "tiaoalil savea",
    "yam banks": "yamarus banks",
    "benjamin yurosek": "ben yurosek",
    "dez malone": "dezjhon malone",
    "josh celiscar": "joshua celiscar",
}
portal["name_final"] = portal["name_clean"].replace(accepted)
```

```python
m3 = portal.merge(wide, left_on=["name_final", "destination"],
                  right_on=["name_clean", "team"], how="left", indicator=True)
dest3 = m3[m3["destination"].isin(wide["team"].unique())]
print(dest3["_merge"].value_counts())
```
